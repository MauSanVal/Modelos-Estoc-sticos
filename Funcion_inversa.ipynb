```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def analizar_muestra(muestra, nombre):
    media = np.mean(muestra)
    varianza = np.var(muestra)
    desviacion = np.std(muestra)

    print(f"Distribución: {nombre}")
    print(f"Media: {media}")
    print(f"Varianza: {varianza}")
    print(f"Desviación estándar: {desviacion}")

    # Histograma
    plt.figure(figsize=(10,4))
    plt.hist(muestra, bins=50, density=True)
    plt.title(f"Histograma - {nombre}")
    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.show()

    # Muestra ordenada
    muestra_ordenada = np.sort(muestra)
    F_emp = np.arange(1, len(muestra)+1)/len(muestra)

    plt.figure(figsize=(10,4))
    plt.plot(muestra_ordenada, F_emp)
    plt.title(f"Función acumulada empírica - {nombre}")
    plt.xlabel("x")
    plt.ylabel("F(x)")
    plt.show()
```

# Distribución Weibull

La función de distribución acumulada es:

$$
F(x)=1-e^{-x^\alpha}
$$

Aplicando transformada inversa:

$$
U=1-e^{-x^\alpha}
$$

Despejando:

$$
x=(-\ln U)^{1/\alpha}
$$

```python
n = 10000
alpha = 2

U = np.random.uniform(0,1,n)
X = (-np.log(U))**(1/alpha)

analizar_muestra(X, "Weibull")
```

# Distribución Gumbel

CDF:

$$
F(x)=e^{-e^{-x}}
$$

Transformada inversa:

$$
x=-\ln(-\ln U)
$$

```python
n = 10000

U = np.random.uniform(0,1,n)
X = -np.log(-np.log(U))

analizar_muestra(X, "Gumbel")
```

# Distribución Cauchy

CDF:

$$
F(x)=\frac{1}{\pi}\arctan(x)+\frac12
$$

Transformada inversa:

$$
x=\tan\left(\pi(U-\frac12)\right)
$$

```python
n = 10000

U = np.random.uniform(0,1,n)
X = np.tan(np.pi*(U-0.5))

analizar_muestra(X, "Cauchy")
```

# Distribución Logística

CDF:

$$
F(x)=\frac{1}{1+e^{-x}}
$$

Transformada inversa:

$$
x=\ln\left(\frac{U}{1-U}\right)
$$

```python
n = 10000

U = np.random.uniform(0,1,n)
X = np.log(U/(1-U))

analizar_muestra(X, "Logística")
```

# Distribución Pareto

CDF:

$$
F(x)=1-x^{-\alpha}
$$

Transformada inversa:

$$
x=(1-U)^{-1/\alpha}
$$

```python
n = 10000
alpha = 3

U = np.random.uniform(0,1,n)
X = (1-U)**(-1/alpha)

analizar_muestra(X, "Pareto")
```
